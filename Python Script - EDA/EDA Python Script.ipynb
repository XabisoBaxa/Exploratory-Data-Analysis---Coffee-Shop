```python
#Creating a new column called Revenue
df['Revenue'] = df['transaction_qty'] * df['unit_price']
df
```

```python
# Convert columns to date types
df['transaction_date'] = pd.to_datetime(df['transaction_date'])
df['transaction_time'] = pd.to_datetime(df['transaction_time'])
df.dtypes
```

```python
#Creating new columns extracting from existing ones 
df['hour'] = df['transaction_time'].dt.hour
df['day'] = df['transaction_date'].dt.day_name()
df['month'] = df['transaction_date'].dt.month_name()
df['year'] = df['transaction_date'].dt.year
df.head()
```

```python
# Sales by store location 
Location_sales = df.groupby('store_location')['Revenue'].sum()
Location_sales
```

### Visualization

```python
import matplotlib.pyplot as plt
```

```python
### Line graph that show sales by date
Sales_by_date = df.groupby("transaction_date")["Revenue"].sum()
Sales_by_date
```

```python
Sales_by_date.plot.line()
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]
plt.plot(x,y)
plt.title("Square Numbers")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
# Bar Graph that shows sales by store location
Sales_by_location = df.groupby("store_location")["Revenue"].sum()
Sales_by_location.plot.bar()
plt.title("Sales by Store Location")
plt.xlabel("Store Location")
plt.ylabel("Revenue")
plt.show()
```

```python
# Pie Chart that shows sales by store location
Sales_by_location = df.groupby("store_location")["Revenue"].sum()
Sales_by_location.plot.pie()
plt.title("Sales by Store Location")
plt.xlabel("Store Location")
plt.ylabel("Revenue")
```
